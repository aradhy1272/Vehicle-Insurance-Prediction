### Split the column 


```python
# spliting  max_torque columns
df[['torque','rpm']]=df['max_torque'].str.extract(r'(\d+)\D+(\d+)')
df[['torque','rpm']]=df[['torque','rpm']].astype(float)
```

```python
# spliting  max_power columns
df[['power','power_rpm']]=df['max_power'].str.extract(r'(\d+)\D+(\d+)')
df[['power','power_rpm']]=df[['power','power_rpm']].astype(float)
```

```python
# extracting   numrical  values from length and width
df['Length'] = df['Length'].str.extract(r'(\d+)').astype(float)
df['width']  = df['width'].str.extract(r'(\d+)').astype(float)
```

```python
# extracting   numrical  values Gross_weight
df['Gross_weight']  = df['Gross_weight'].str.extract(r'(\d+)').astype(float)
```

```python
# Drop columns
df.drop(['max_torque','max_power'],axis=1,inplace=True)
```

```python
# Split Data
x=df.drop(['make','is_claim'],axis=1)
y=df['is_claim']
```

```python
# Split Data using train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=12345)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
x_train_res, y_train_res = smote.fit_resample(x_train, y_train)
```

# Building Model

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegressionCV

def create_logistic_regression_model():
    model=LogisticRegressionCV()

    model.fit(x_train_res,y_train_res)

    return model
```

### KNN

```python
def create_knn_model():
    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.model_selection import GridSearchCV
    model=KNeighborsClassifier(n_neighbors=100)
    model.fit(x_train_res,y_train_res)
    return model
```

### SVM

```python

from sklearn.svm import SVC
def create_svm_model():
    model=SVC()
    model.fit(x_train_res,y_train_res)
    return model
```

###  DECISION TREE

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
def create_decision_tree_model():
    model=DecisionTreeClassifier()
    parameters={
      'criterion':['gini','log_loss','entropy'],
      'max_depth':[5,10,15,20]
    }
   
    model_cv=GridSearchCV(model,parameters)
    model_cv.fit(x_train_res,y_train_res)
    best_model_tree=model_cv.best_estimator_
    return  best_model_tree 
```

# Model Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,roc_auc_score

def evaluate_model(model):


    # testing accuracy
    y_pred=model.predict(x_test)
    testing_accuracy=accuracy_score(y_test,y_pred)
    testing_precision=precision_score(y_test,y_pred)
    testing_recall=recall_score(y_test,y_pred)
    testing_f1=f1_score(y_test,y_pred)
    testing_roc = roc_auc_score(y_test, y_pred)
    return  testing_accuracy, testing_precision, testing_recall, testing_f1, testing_roc
```

## Generate  Performance chart

```python
models=[
("Logistic Regression",create_logistic_regression_model(),(0,0)),
("KNN",create_knn_model(),(0,1)),
("SVM",create_svm_model(),(1,0)),
("Decision Tree",create_decision_tree_model(),(1,1)),
]
```

```python
performance_data = []
for model_name, model, position in models:
    test_accuracy, precision, recall, f1, roc = evaluate_model(model)
    performance_data.append([
        model_name, f"{test_accuracy * 100:0.2f}%",f"{precision * 100:0.2f}%", f"{recall * 100:0.2f}%",  f"{f1 * 100:0.2f}%", f"{roc * 100:0.2f}%"
    ])
    
performance_chart = pd.DataFrame(performance_data, 
            columns=["Model","Test Accuracy", "Precision", "Recall", "F1", "ROC"])
performance_chart
```
